```python
import pandas as pd
import requests
import os

os.makedirs("../data/raw/internal", exist_ok=True)
os.makedirs("../data/raw/open_data", exist_ok=True)

file_path = "../data/raw/internal/Factures.csv"

df = pd.read_csv(
    file_path,
    sep=";",
    low_memory=False
)

df.head()
```

```python
# Sélection des colonnes utiles

df_internal = df[
    [
        "Unnamed: 0",
        "date_invoiced",
        "market_level1",
        "market_level2",
        "market_level3",
        "qty",
        "trans_amount_weight",
        "geo_region",
        "cust_country",
        "cust_country_code",
        "Currency_corp"
    ]
].copy()

df_internal.head()
```

```python
# Renommage des colonnes

df_internal = df_internal.rename(columns={
    "Unnamed: 0": "transaction_id",
    "date_invoiced": "date_vente",
    "market_level1": "categorie_produit_1",
    "market_level2": "categorie_produit_2",
    "market_level3": "produit",
    "qty": "quantite_vendue",
    "geo_region": "region",
    "cust_country": "pays",
    "cust_country_code": "code_pays",
    "Currency_corp": "devise"
})

df_internal["date_vente"] = pd.to_datetime(
    df_internal["date_vente"],
    errors="coerce"
)

df_internal["quantite_vendue"] = pd.to_numeric(
    df_internal["quantite_vendue"],
    errors="coerce"
)

df_internal.head()
```

```python
# Filtre sur l'agriculture en Amérique du Nord

df_internal = df_internal[
    (
        df_internal["categorie_produit_1"]
        .astype(str)
        .str.strip()
        .str.lower()
        == "agriculture"
    )
    &
    (
        df_internal["region"]
        .astype(str)
        .str.strip()
        .str.lower()
        == "northern america"
    )
    &
    (
        df_internal["code_pays"]
        .astype(str)
        .str.strip()
        .str.upper()
        .isin(["USA", "CAN"])
    )
].copy()

df_internal.head()
```

```python
# Informations après filtrage
df_internal.info()
```

```python
# Valeurs manquantes après filtrage
df_internal.isnull().sum()
```

```python
# Dimensions après filtrage
df_internal.shape
```

```python
# Vérification de l'identifiant des transactions

print(
    "Identifiants manquants :",
    df_internal["transaction_id"].isna().sum()
)

print(
    "Identifiants dupliqués :",
    df_internal["transaction_id"].duplicated().sum()
)

# Sauvegarde des ventes filtrées

df_internal.to_csv(
    "../data/raw/internal/ventes_brutes.csv",
    index=False
)

date_debut = df_internal["date_vente"].min()
date_fin = df_internal["date_vente"].max()

start_year = date_debut.year
end_year = date_fin.year

start_date = f"{start_year}-01-01"
end_date = f"{end_year}-12-31"

print("Nombre de lignes :", len(df_internal))
print("Période :", date_debut, "au", date_fin)
print("Pays :", df_internal["code_pays"].unique())
```

```python
# Données météo des États-Unis et du Canada

url = "https://archive-api.open-meteo.com/v1/archive"

pays_meteo = {
    "USA": [39.50, -98.35],
    "CAN": [52.00, -106.00]
}

weather = []

for code_pays, coordonnees in pays_meteo.items():

    params = {
        "latitude": coordonnees[0],
        "longitude": coordonnees[1],
        "start_date": start_date,
        "end_date": end_date,
        "daily": "temperature_2m_mean,precipitation_sum",
        "timezone": "UTC"
    }

    response = requests.get(
        url,
        params=params,
        timeout=120
    )

    response.raise_for_status()

    data = response.json()["daily"]

    weather_pays = pd.DataFrame({
        "date": data["time"],
        "temperature": data["temperature_2m_mean"],
        "precipitation": data["precipitation_sum"]
    })

    weather_pays["code_pays"] = code_pays

    weather.append(weather_pays)

weather = pd.concat(
    weather,
    ignore_index=True
)

weather.head()
```

```python
# Sauvegarde des données météo

weather.to_csv(
    "../data/raw/open_data/weather.csv",
    index=False
)

print("Dimensions météo :", weather.shape)
print("Colonnes :", weather.columns.tolist())
print(weather["code_pays"].value_counts())
```

```python
# Inflation annuelle - Banque Mondiale

url_inflation = (
    "https://api.worldbank.org/v2/country/USA;CAN/"
    "indicator/FP.CPI.TOTL.ZG"
    f"?format=json&date={start_year}:{end_year}&per_page=20000"
)

response = requests.get(
    url_inflation,
    timeout=60
)

response.raise_for_status()

data = response.json()[1]

inflation = pd.DataFrame(data)

inflation = inflation[
    [
        "countryiso3code",
        "date",
        "value"
    ]
]

inflation = inflation.rename(columns={
    "countryiso3code": "code_pays",
    "value": "inflation"
})

inflation["inflation"] = pd.to_numeric(
    inflation["inflation"],
    errors="coerce"
)

inflation = inflation.dropna()

inflation = inflation.sort_values(
    ["code_pays", "date"]
)

inflation.head()
```

```python
# Sauvegarde de l’inflation

inflation.to_csv(
    "../data/raw/open_data/inflation.csv",
    index=False
)

print("Dimensions inflation :", inflation.shape)
```

```python
# Croissance annuelle du PIB - Banque Mondiale

url_gdp = (
    "https://api.worldbank.org/v2/country/USA;CAN/"
    "indicator/NY.GDP.MKTP.KD.ZG"
    f"?format=json&date={start_year}:{end_year}&per_page=20000"
)

response = requests.get(
    url_gdp,
    timeout=60
)

response.raise_for_status()

data = response.json()[1]

gdp = pd.DataFrame(data)

gdp = gdp[
    [
        "countryiso3code",
        "date",
        "value"
    ]
]

gdp = gdp.rename(columns={
    "countryiso3code": "code_pays",
    "value": "gdp_growth"
})

gdp["gdp_growth"] = pd.to_numeric(
    gdp["gdp_growth"],
    errors="coerce"
)

gdp = gdp.dropna()

gdp = gdp.sort_values(
    ["code_pays", "date"]
)

gdp.head()
```

```python
# Sauvegarde du PIB
gdp.to_csv(
    "../data/raw/open_data/gdp.csv",
    index=False
)

print("Dimensions PIB :", gdp.shape)
```

```python
# Taux d'intérêt mensuels - FRED

interest_series = {
    "USA": "IR3TIB01USM156N",
    "CAN": "IR3TIB01CAM156N"
}

interest_rate = []

for code_pays, serie in interest_series.items():

    url_interest = (
        "https://fred.stlouisfed.org/graph/"
        f"fredgraph.csv?id={serie}"
    )

    interest_pays = pd.read_csv(url_interest)

    interest_pays = interest_pays.rename(columns={
        "observation_date": "date",
        serie: "interest_rate"
    })

    interest_pays["date"] = pd.to_datetime(
        interest_pays["date"],
        errors="coerce"
    )

    interest_pays["interest_rate"] = pd.to_numeric(
        interest_pays["interest_rate"],
        errors="coerce"
    )

    interest_pays = interest_pays[
        (interest_pays["date"].dt.year >= start_year)
        &
        (interest_pays["date"].dt.year <= end_year)
    ]

    interest_pays = interest_pays.dropna()

    interest_pays["code_pays"] = code_pays

    interest_rate.append(interest_pays)

interest_rate = pd.concat(
    interest_rate,
    ignore_index=True
)

interest_rate = interest_rate[
    [
        "date",
        "code_pays",
        "interest_rate"
    ]
]

interest_rate.head()
```

```python
# Sauvegarde des taux d’intérêt
interest_rate.to_csv(
    "../data/raw/open_data/interest_rate.csv",
    index=False
)

print("Dimensions taux d'intérêt :", interest_rate.shape)
print(interest_rate["code_pays"].value_counts())
```

```python
# Collecte du prix du blé
# Prix mondial mensuel du blé - FRED

url_wheat = (
    "https://fred.stlouisfed.org/graph/"
    "fredgraph.csv?id=PWHEAMTUSDM"
)

wheat_price = pd.read_csv(url_wheat)

wheat_price = wheat_price.rename(columns={
    "observation_date": "date",
    "PWHEAMTUSDM": "wheat_price"
})

wheat_price["date"] = pd.to_datetime(
    wheat_price["date"],
    errors="coerce"
)

wheat_price["wheat_price"] = pd.to_numeric(
    wheat_price["wheat_price"],
    errors="coerce"
)

wheat_price = wheat_price[
    (wheat_price["date"].dt.year >= start_year)
    &
    (wheat_price["date"].dt.year <= end_year)
]

wheat_price = wheat_price.dropna()

wheat_price.head()
```

```python
# Sauvegarde du prix du blé
wheat_price.to_csv(
    "../data/raw/open_data/wheat_price.csv",
    index=False
)

print("Dimensions prix du blé :", wheat_price.shape)
```

```python
# Sauvegarde du prix du maïs

corn_price.to_csv(
    "../data/raw/open_data/corn_price.csv",
    index=False
)

print("Dimensions prix du maïs :", corn_price.shape)
```

```python
# Collecte du prix du soja
# Prix mondial mensuel du soja - FRED

url_soybean = (
    "https://fred.stlouisfed.org/graph/"
    "fredgraph.csv?id=PSOYBUSDM"
)

soybean_price = pd.read_csv(url_soybean)

soybean_price = soybean_price.rename(columns={
    "observation_date": "date",
    "PSOYBUSDM": "soybean_price"
})

soybean_price["date"] = pd.to_datetime(
    soybean_price["date"],
    errors="coerce"
)

soybean_price["soybean_price"] = pd.to_numeric(
    soybean_price["soybean_price"],
    errors="coerce"
)

soybean_price = soybean_price[
    (soybean_price["date"].dt.year >= start_year)
    &
    (soybean_price["date"].dt.year <= end_year)
]

soybean_price = soybean_price.dropna()

soybean_price.head()
```
